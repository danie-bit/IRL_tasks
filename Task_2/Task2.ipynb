```python
# Question 1

class Solution:
    def defangIPaddr(self, address: str) -> str:
        out_str = ""
        for i in range(len(address)) :
            if address[i] == ".":
                out_str = out_str + "[.]"
            else:
                out_str += address[i]
        return out_str
```

```python
q1 = Solution()
q1.defangIPaddr("255.255.255.128")
```

```python
# Question 2

from typing import List

class Solution:
    def findNumbers(self, nums: List[int]) -> int:
        cnt = 0
        for i in nums:
            cnt_in = 0
            while i > 0:
                i //= 10
                cnt_in += 1
            if cnt_in % 2 == 0:
                cnt += 1
        return cnt
```

```python
q2 = Solution()
q2.findNumbers([12,345,2,16,7896])
```

```python
# Question 3

class Solution:
    def numIdenticalPairs(self, nums: List[int]) -> int:
        cnt = 0
        for i in range(len(nums)) :
            for j in range(i+1,len(nums)):
                if nums[i] == nums[j]:
                    cnt += 1
        return cnt
```

```python
q3 = Solution()
q3.numIdenticalPairs([1,2,3,1,1,3])
```

```python
# Question 4

class Solution:
    def smallerNumbersThanCurrent(self, nums: List[int]) -> List[int]:
        cnt = []
        for i in range(len(nums)):
            cnt_in = 0
            for j in range(len(nums)):
                if i != j and nums[j]< nums[i]:
                    cnt_in += 1
            cnt.append(cnt_in)
        return cnt
```

```python
q4 = Solution()
q4.smallerNumbersThanCurrent([8,1,2,2,3])
```

```python
# Question 5

class Solution:
    def subtractProductAndSum(self, n: int) -> int:
        sum_ = 0
        prod = 1
        while n > 0 :
            sum_ += (n % 10)
            prod *= (n % 10)
            n //= 10
        return prod - sum_
```

```python
q5 = Solution()
q5.subtractProductAndSum(674)
```

```python
# Question 6

class Solution:
    def xorOperation(self, n: int, start: int) -> int:
        nums = []
        for i in range(n):
            nums.append((2*i) + start)
        xor_ = nums[0]
        for i in range(n-1):
            xor_ ^= nums[i+1]
        return xor_
```

```python
q6 = Solution()
q6.xorOperation(4,3)
```
